```python
#Question 6
m1 = df['horsepower'].median()
m2 = df['horsepower'].mode()[0]

df1 = df['horsepower'].fillna(m2)
m3 = df1.median()

result = (m3/m1)-1

if result > 0:
    print(f"The median increased in {result*100:.2f}%")
elif result == 0:
    print(f"The median stayed the same")
else:
    print(f"The median decreased in {result*100:.2f}%")
```

```python
#Question 7
df2 = df[df['origin'] == 'Asia']
x = np.array(df2[['vehicle_weight','model_year']].head(7))
xtx = x.T.dot(x)
inv_xtx = np.linalg.inv(xtx)
y = np.array([1100, 1300, 800, 900, 1000, 1100, 1200])
w = inv_xtx.dot(x.T).dot(y)
f7 = w.sum()
print(f"The sum of the weights is {f7:.3f}")
```
